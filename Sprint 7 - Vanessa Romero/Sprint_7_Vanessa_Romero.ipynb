### 🧩 Desarrollo de funciones

⚡`obtener_datos()`

1. Pedir nombre y apallido y almacena
2. Pedir número de teléfiono y almacenar
3. Pedir años de experiencia y almacenar
4. Devuelve nombre completo, teléfono y experiencia

```python
def obtener_datos():
    nombre_completo = input("Introduce tu primer nombre y primer apellido: ")
    telefono = input("Introduce un teléfono de contacto: ")
    while True:
      experiencia = input("Introduce tus años de experiencia: ")
      experiencia = validar_experiencia(experiencia)
      if experiencia is not None:
          break
    return nombre_completo, telefono, experiencia
```

⚡`validar_experiencia()`

1. Confirmar la experiencia es un número válido para los calculos posteriores
2. Mostrar mensaje de error si falla algún dato.
3. Almacenar datos correctos.


```python
def validar_experiencia(valor):
    valor = valor.replace(",", ".")
    try:
        valor = float(valor)
        if valor >= 0:
            return valor
        else:
            print("Error: La experiencia no puede ser negativa.")
            return None
    except ValueError:
        print("Error: Por favor, introduce números válidos.")
        return None
```

```python
# Prueba obtener_datos + validar_experiencia
datos = obtener_datos()
```

⚡`nivel_cat()`

1. Mostrar al usuario las opciones de nivel.
2. Preguntar al usuario su nivel.
3. Almacenar la respuesta en catalan.

```python
def nivel_cat():
    print("Nivel de Catalán:")
    print("  0 = Sin conocimientos")
    print("  1 = Hasta el Nivel B1")
    print("  2 = Nivel B2")
    print("  3 = Nivel C1 o superior")

    catalan = input("Introduce tu Nivel de Catalán (0-3): ")
    return validar_nivel(catalan, 0, 3)
```

⚡`nivel_excel()`

1. Mostrar al usuario las opciones de nivel.
2. Preguntar al usuario su nivel.
3. Almacenar la respuesta en excel.

```python
def nivel_excel():
    print("Nivel de Excel:")
    print("  0 = Nivel básico")
    print("  1 = Uso de funciones")
    print("  2 = Tablas dinámicas")
    print("  3 = Visual Basic y Macros")

    excel = input("Introduce tu Nivel de Excel (0-3): ")
    return validar_nivel(excel, 0, 3)
```

⚡`nivel_estudios()`

1. Mostrar al usuario las opciones de nivel.
2. Preguntar al usuario su nivel.
3. Almacenar la respuesta en estudios.

```python
def nivel_estudios():
    print("Título Académico:")
    print("  0 = ESO")
    print("  1 = FP")
    print("  2 = Universidad")
    print("  3 = Máster o Superior")

    estudios = input("Introduce tu Título Académico (0-3): ")
    return validar_nivel(estudios, 0, 3)
```

⚡`validar_nivel()`

1. Confirmar que el número cumpla las condiciones:
    *   Sea entero
    *   Esté entre 0 y 3
2. Almacenar datos correctos.

```python
def validar_nivel(valor, minimo, maximo):
    try:
        valor = int(valor)
        if minimo <= valor <= maximo:
            return valor
        else:
            return None
    except ValueError:
        return None
```

```python
# Prueba nivel_cat, nivel_excel, nivel_estudios + validar_nivel
catalan = nivel_cat()
excel = nivel_excel()
estudios = nivel_estudios()
```

⚡`calcular_puntuacion()`

1. Multiplicar los años de experiencia por 1.5.
2. Sumar los puntos de los distintos niveles evaluados.
3. Almacenar el resultado total como `float`.

```python
def calcular_puntuacion(valor, num1, num2, num3):
    experiencia = valor * 1.5
    conocimiento = num1 + num2 + num3
    puntuacion = experiencia + conocimiento
    return float(puntuacion)
```

```python
# Prueba calcular_puntuacion
puntuacion = calcular_puntuacion(7, 0, 2, 3)
print(puntuacion)
```

⚡`asignar_clasificacion()`

1. Asignar valores a cada rango de puntuacion.
2. Almacenar resultados por grupo.

```python
def asignar_clasificacion(valor):
    match valor:
        case alto if alto >= 12:
            return "Prioridad Alta"
        case medio if 8 <= medio < 12:
            return "Candidato/a adecuado/a"
        case bajo:
            return "No prioritario/a"
```

```python
# Prueba asignar_clasificacion
clasificacion = asignar_clasificacion(15.5)
print(clasificacion)
```

⚡`mostrar_resultados()`

1. Mostrar al usuario la puntuación obtenida.

```python
def mostrar_resultados(nombre_completo, valor):
    print(f"Gracias por inscribirte {nombre_completo}, tu puntuación como candidato es {valor}")
    print("\n")
```

```python
# Prueba mostrar_resultados
resultados = mostrar_resultados("Vanessa Romero", 15.5)
```

⚡`lista_candidatos()`

1. Crear una lista con la información de los candidatos
2. Bloquear lista de candidatos cuando llegue a 10
3. Añadir cada diccionario a la lista candidatos


```python
def lista_candidatos():
    candidatos = []

    while len(candidatos) < 10:
        nombre_completo, telefono, experiencia = obtener_datos()
        cat = nivel_cat()
        exc = nivel_excel()
        est = nivel_estudios()

        if None in (experiencia, cat, exc, est):
            print("Registro descartado por datos incorrectos. Vuelve a intentarlo")
            print("\n")
            continue

        puntuacion = calcular_puntuacion(experiencia, cat, exc, est)
        clasificacion = asignar_clasificacion(puntuacion)
        mostrar_resultados(nombre_completo, puntuacion)

        candidato = {
            "nombre": nombre_completo,
            "telefono": telefono,
            "puntuacion": puntuacion,
            "clasificacion": clasificacion
        }
        candidatos.append(candidato)

    print("\n🚫 Inscripciones cerradas. Se han registrado 10 candidatos.")

    return candidatos
```

⚡`generar_informe()`

1. Ordenar lista de candidatos
2. Mostrar los candidatos ordenados


```python
def generar_informe(candidatos):
    orden_prioridad = {
        "Prioridad Alta": 1,
        "Candidato/a adecuado/a": 2,
        "No prioritario/a": 3
    }

    candidatos_ordenados = sorted(
        candidatos,
        key=lambda c: (orden_prioridad[c["clasificacion"]], -c["puntuacion"])
    )

    print("\n===== INFORME FINAL DE CANDIDATOS =====")
    for candidato in candidatos_ordenados:
        print(f"Nombre: {candidato['nombre']}")
        print(f"Teléfono: {candidato['telefono']}")
        print(f"Puntuación: {candidato['puntuacion']}")
        print(f"Clasificación: {candidato['clasificacion']}")
        print("---------------------------------------")
```

### 🏆 Resultado final (INTEGRAR)

```python
# Creación de lista de candidatos y generación de informe final
resultado_seleccion_person = generar_informe(lista_candidatos())
```

⚡`obtener_segmento_factor()`

1. Recibir el código de comportamiento de un cliente.
2. Asignar factor de comportamiento asociado.
3. Asignar mensaje asociado.
4. Devolver factor y mensaje.

```python
def obtener_segmento_factor(codigo_comportamiento):
    segmento = {
        0: {"factor": 0.10, "missatge": "Navega però no afegeix productes"},
        1: {"factor": 0.25, "missatge": "Afegeix productes però no arriba al checkout"},
        2: {"factor": 0.50, "missatge": "Arriba al checkout però no paga"},
        3: {"factor": 0.75, "missatge": "Cistella abandonat fa menys de 24h"},
        4: {"factor": 0.90, "missatge": "Cistella abandonat fa més de 24h"}
    }

    return segmento[codigo_comportamiento]
```

```python
# Prueba obtener_segmento_factor
segmento = obtener_segmento_factor(0)
print(segmento)
```

⚡`obtener_factor_edad()`

1. Recibir el grupo de edad del cliente
2. Asignar factor de edad asociado
3. Devolver factor

```python
def obtener_factor_edad(grupo_edad):
    factor = {
        "18-25": 1.20,
        "26-40": 1.00,
        "41-60": 0.85,
        "60+": 0.70
    }
    return factor[grupo_edad]
```

```python
# Prueba obtener_factor_edad
factor_edad = obtener_factor_edad("41-60")
print(factor_edad)
```

⚡`calcular_riesgo()`

1. Recibir el factor de comportamiento del cliente
2. Recibir el factor de edad del cliente
3. Aplicar la operación matemática.
4. Devolver el valor de riesgo.

```python
def calcular_riesgo(comportamiento, edad):
  calculo = float(comportamiento) * float(edad)
  return round(calculo, 2)
```

```python
# Prueba calcular_riesgo
riesgo = calcular_riesgo(0.1, 0.85)
print(riesgo)
```

⚡`procesar_clientes()`

1. Recibir datos del cliente.
2. Asignar código de comportamiento y mensaje.
3. Asignar factor de edad.
4. Hacer calculo de riesgo.

```python
def procesar_clientes(lista_clientes):
    resultado = {}

    for cliente in lista_clientes:
        segmento = obtener_segmento_factor(cliente["codi_comportament"])
        factor_edad = obtener_factor_edad(cliente["grup_edat"])
        riesgo = calcular_riesgo(segmento["factor"], factor_edad)

        resultado[cliente["id_client"]] = {
            "nom": cliente["nom"],
            "edat": cliente["edat"],
            "grup_edat": cliente["grup_edat"],
            "codi_comportament": cliente["codi_comportament"],
            "missatge": segmento["missatge"],
            "factor_codi": segmento["factor"],
            "factor_edat": factor_edad,
            "risc_final": riesgo
        }

    return resultado
```

### 🧩 Desarrollo de funciones

⚡`separar_pacientes()`

1. Recibe string
2. Asigna claves especificas
3. Asigna valores a cada clave
4. Devuelve diccionario

```python
def separar_pacientes (string):

  claves = ["Nombre", "Edad", "Síntomas"]

  paciente = dict(zip(claves, string.split("|")))

  return paciente
```

```python
#Prueba de uso
paciente_1 = separar_pacientes ("Maria|45|tos, Fiebre Alta, dolor_pecho")

print(paciente_1)
```

⚡`estandarizar_sintomas()`

1. Recibe diccionario
2. Convierte todas las letras en minúsculas
3. Sustituye espacios por "under_score"
4. Devuelve el diccionario con la lista de síntomas limpia y estandarizada

```python
def estandarizar_sintomas(diccionario):

    sintomas = diccionario["Síntomas"].split(",")

    sintomas_limpios = []

    for sintoma in sintomas:
        sintoma = sintoma.strip()
        sintoma = sintoma.lower()
        sintoma = sintoma.replace(" ", "_")

        sintomas_limpios.append(sintoma)

    diccionario["Síntomas"] = sintomas_limpios

    return diccionario
```

```python
#Prueba de uso
paciente_2 = estandarizar_sintomas (paciente_1)

print(paciente_2)
```

⚡`calcular_riesgo_paciente()`

1. Recibe diccionario.
2. Suma valores númericos si hay coincidencia de clave en el diccionario base.
3. Devuelve suma de las coincidencias.

```python
def calcular_riesgo_paciente (diccionario_paciente):

  nivells_risc = {
    "fiebre_alta": 3,
    "dificultad_respirar": 5,
    "dolor_pecho": 4,
    "tos": 1,
    "fatiga": 1,
    "dolor_cabeza": 1
  }

  clave_sintoma = diccionario_paciente["Síntomas"]
  riesgo_total = 0

  for sintoma in clave_sintoma:
    if sintoma in nivells_risc:
        riesgo_total = riesgo_total + nivells_risc[sintoma]

  return riesgo_total
```

```python
#Prueba de uso
prueba_maria = {
    'Nombre': 'Maria',
    'Edad': '45',
    'Síntomas': ['tos', 'fiebre_alta', 'dolor_pecho']
}

riesgo_maria = calcular_riesgo_paciente(prueba_maria)

print(riesgo_maria)
```

⚡`crear_paciente()`

1. Recibe diccionario.
2. Añade nueva clave y valor al diccionario.
3. Devuelve diccionario actualizado.

```python
def crear_paciente (diccionario_inicial):

  diccionario_inicial ["Riesgo"] = calcular_riesgo_paciente(diccionario_inicial)

  return diccionario_inicial
```

```python
#Prueba de uso
paciente_maria = {
    'Nombre': 'Maria',
    'Edad': '45',
    'Síntomas': ['tos', 'fiebre_alta', 'dolor_pecho']
}

crear_maria = crear_paciente (paciente_maria)

print(crear_maria)
```

⚡`ordenar_riesgo()`

1. Recibe lista de diccionarios.
2. Ordena la lista por el valor de una clave específica de mayor a menor.
3. Devuelve lista de diccionarios ordenada.

```python
def ordenar_riesgo(lista_pacientes):

    return sorted(
        lista_pacientes,
        key=lambda x: x["Riesgo"],
        reverse=True
    )
```

```python
#Prueba de uso
prueba_pacientes = [
    {"Nombre": "Maria", "Riesgo": 8},
    {"Nombre": "Luis", "Riesgo": 2},
    {"Nombre": "Sara", "Riesgo": 9}
]

resultado = ordenar_riesgo(prueba_pacientes)

print(resultado)
```

⚡`main_pacientes()`

1. Recibe una lista sin formato.
2. Aplica las funciones creadas.
3. Devuelve un DataFrame con los pacientes ordenados por nivel de riesgo.

```python
import pandas as pd
```

```python
def main_pacientes(lista_inicial):

    lista_pacientes = []

    for paciente in lista_inicial:

        paciente_separado = separar_pacientes(paciente)
        paciente_limpio = estandarizar_sintomas(paciente_separado)
        paciente_final = crear_paciente(paciente_limpio)

        lista_pacientes.append(paciente_final)

    pacientes_ordenados = ordenar_riesgo(lista_pacientes)

    return pd.DataFrame(pacientes_ordenados)
```

```python
#Prueba de uso
pacientes_prueba = [
    "Carlos|40|Tos, dolor_pecho",
    "Lucia|25|fatiga, Dolor Cabeza",
    "Pedro|70|Fiebre Alta, dificultad RESPIRAR"
]

prueba_main = main_pacientes(pacientes_prueba)

print(prueba_main)
```

### 🏆 Resultado final

```python
# Ejecución de main_pacientes con datos reales del ejercicio

pacients = [

  "Maria|45|tos, Fiebre Alta, dolor_pecho",
  "Luis|33|dolor_cabeza,faTiga",
  "Sara|67|fiebre alta, dificultad RESPIRAR, tos",
  "Jordi|52|fatiga , tos",
  "Anna|29|fiebre alta,dolor_cabeza"
]

resultado_final = main_pacientes(pacients)

print(resultado_final)
```

### 🧩 Desarrollo de funciones

⚡`add_nivells()`

1. Recibe un diccionario.
2. Compara con diccionario interno.
3. Asigna claves especificas.
4. Asigna valores a cada clave.
5. Devuelve diccionario.

```python
def add_nivells (diccionario_inicial):

  nivells_seguro = {

    0: {"categoria": "Baix risc", "preu_base": 120, "fraude": 0.01},
    1: {"categoria": "Risc mitjà", "preu_base": 200, "fraude": 0.05},
    2: {"categoria": "Risc alt", "preu_base": 350, "fraude": 0.15},
    3: {"categoria": "Risc crític", "preu_base": 500, "fraude": 0.30}
  }

  nivell_actual = diccionario_inicial["codi"]

  diccionario_inicial.update(nivells_seguro[nivell_actual])

  return diccionario_inicial
```

```python
#Prueba de uso
cliente_prueba = {
    "nom": "Lucía",
    "edat": 37,
    "codi": 2
}

prueba_uso_01 = add_nivells(cliente_prueba)
print(prueba_uso_01)
```

⚡`calcular_preu()`

1. Recibe diccionario
2. Aplica fórmula matemática si hay coincidencia de claves en el diccionario base
3. Devuelve el resultado de la fórmula

```python
def calcular_preu (diccionario_cliente):

  clave_preu = diccionario_cliente["preu_base"]
  clave_edat = diccionario_cliente["edat"]

  preu_poliza = clave_preu * (1 + clave_edat / 100)

  return round(preu_poliza, 2)
```

```python
#Prueba de uso
prueba_02 = {
    "nom": "Lucía",
    "edat": 37,
    "codi": 2,
    "categoria": "Risc alt",
    "preu_base": 350,
    "fraude": 0.15
}

prueba_uso_02 = calcular_preu(prueba_02)
print(prueba_uso_02)
```

⚡`add_preu()`

1. Recibe diccionario.
2. Añade nueva clave y valor al diccionario.
3. Devuelve diccionario actualizado.

```python
def add_preu (diccionario_incompleto):

  diccionario_incompleto ["preu_final"] = calcular_preu(diccionario_incompleto)

  return diccionario_incompleto
```

```python
#Prueba de uso
cliente_x = {
    "nom": "Marc",
    "edat": 45,
    "codi": 1,
    "categoria": "Risc mitjà",
    "preu_base": 200,
    "fraude": 0.05
}

prueba_uso_03 = add_preu (cliente_x)

print(prueba_uso_03)
```

⚡`ordenar_clients()`

1. Recibe lista de diccionarios
2. Ordena por fraude descendente
3. En caso de empate, ordena por preu_final ascendente
4. Devuelve un DataFrame con los clientes ordenados

```python
def ordenar_clients(lista_clients):

    clients_ordenats = sorted(
        lista_clients,
        key=lambda x: (-x["fraude"], x["preu_final"])
    )

    df_clients = pd.DataFrame(clients_ordenats)

    return df_clients
```

```python
#Prueba de uso
clients_04 = [
    {"nom": "Laura", "fraude": 0.15, "preu_final": 420.0},
    {"nom": "Pau", "fraude": 0.30, "preu_final": 610.0},
    {"nom": "Mireia", "fraude": 0.30, "preu_final": 550.0},
    {"nom": "Joan", "fraude": 0.05, "preu_final": 300.0}
]

prueba_uso_04 = ordenar_clients (clients_04)

print(prueba_uso_04)
```

⚡`analitzar_clients()`

1. Recibe un DataFrame.
2. Ejecuta `client_major_fraude()`.
      
    2.1 Recibe un DataFrame.

    2.2 Busca el valor máximo en fraude.
    
    2.3 Devuelve la fila del cliente.
3. Ejecuta `client_major_preu()`.
      
    3.1 Recibe un DataFrame.

    3.2 Busca el valor máximo en preu_final.
    
    3.3 Devuelve la fila del cliente.
4. Ejecuta `client_revisio_manual()`.

    4.1 Recibe un DataFrame.

    4.2 Busca los clientes con "Risc crític" o "Risc alt".
    
    4.3 Devuelve el DataFrame de clientes seleccionados.

5. Devuelve los resultados del análisis.

```python
def client_major_fraude(df_clients):

    index_major_fraude = df_clients["fraude"].idxmax()
    major_fraude = df_clients.loc[index_major_fraude]

    return major_fraude
```

```python
def client_major_preu(df_clients):

    index_major_preu = df_clients["preu_final"].idxmax()
    major_preu = df_clients.loc[index_major_preu]

    return major_preu
```

```python
def client_revisio_manual(df_clients):

    revisio_manual = df_clients[
        df_clients["categoria"].isin(["Risc crític", "Risc alt"])
    ]

    return revisio_manual
```

```python
def analitzar_clients(df_clients):

    major_fraude = client_major_fraude(df_clients)
    major_preu = client_major_preu(df_clients)
    revisio_manual = client_revisio_manual(df_clients)

    print("El client amb major probabilitat de frau és:")
    print(major_fraude)

    print("\nEl client amb el preu final més alt és:")
    print(major_preu)

    print("\nClients prioritaris per revisió manual:")
    print(revisio_manual)

    return major_fraude, major_preu, revisio_manual
```

⚡`main_clients()`

1. Recibe una lista de diccionarios
2. Aplica las funciones creadas
3. Devuelve el DataFrame de clientes ordenados y los resultados del análisis

```python
def main_clients(lista_inicial_clientes):

    lista_clientes_procesados = []

    for cliente in lista_inicial_clientes:

        niveles = add_nivells(cliente)
        preu_final = add_preu(niveles)

        lista_clientes_procesados.append(preu_final)

    clientes_ordenados = ordenar_clients(lista_clientes_procesados)

    print("Clients ordenats:")
    display(clientes_ordenados)

    clientes_analizados = analitzar_clients(clientes_ordenados)

    return clientes_ordenados, clientes_analizados
```

### 🧩 Desarrollo de funciones

⚡`pw_must()`

1. Recibe la longitud mínima que deberá tener la contraseña
2. Pregunta al usuario la longitud de su contraseña
3. Envía mensaje de error si la longitud seleccionada es inferior a la mínima establecida.
4. Pregunta al usuario que tipo de caracteres quiere incluir (True/False).
5. Devuelve length y elementos mínimos de creación.

```python
def pw_must (min_length):

  length = int(input(f"¿Cuántos caracteres quieres para tu contraseña (minimo {min_length})? "))

  while length < min_length:
      print(f"Recuerda que la contraseña debe tener un mínimo de {min_length} caracteres.")
      length = int(input("¿Cuántos caracteres quieres para tu contraseña? "))

  mayuscula = input("¿Quieres mayúsculas? (Si/No) ").strip().lower() in ("si", "sí")
  minuscula = input("¿Quieres minúsculas? (Si/No) ").strip().lower() in ("si", "sí")
  numero = input("¿Quieres números? (Si/No) ").strip().lower() in ("si", "sí")
  simbolo = input("¿Quieres símbolos? (Si/No) ").strip().lower() in ("si", "sí")

  return length, mayuscula, minuscula, numero, simbolo
```

⚡`pw_elements()`

1. Recibe elementos a buscar (True/False)
2. Define un caracter aleatorio por cada elemento
3. Añade cada elemento en formato `string` de Python a un lista
3. Devuelve  lista de caracteres minimos

```python
import string
import numpy as np
```

```python
def pw_elements(mayuscula, minuscula, numero, simbolo):

    p_list = []
    azar = np.random.default_rng()

    if mayuscula:
        mayusculas = string.ascii_uppercase
        e_mayuscula = azar.choice(list(mayusculas))
        p_list.append(str(e_mayuscula))

    if minuscula:
        minusculas = string.ascii_lowercase
        e_minuscula = azar.choice(list(minusculas))
        p_list.append(str(e_minuscula))

    if numero:
        numeros = string.digits
        e_numero = azar.choice(list(numeros))
        p_list.append(str(e_numero))

    if simbolo:
        simbolos = string.punctuation
        e_simbolo = azar.choice(list(simbolos))
        p_list.append(str(e_simbolo))

    return p_list
```

```python
#Prueba de uso

prueba_pw_elements = pw_elements(True, True, False, True)

print(prueba_pw_elements)
```

⚡`pw_length()`

1. Recibe la longitud establecida en `length` y el número de elementos mínimos establecidos (True/False)
2. Calcula la lista de caracteres mínimos generada por `pw_elements()`
3. Devuelve un resultado númerico

```python
def pw_length (length, pw_basic):

  pw_faltantes = length - len(pw_basic)

  return pw_faltantes
```

```python
#Prueba de uso
pw_prueba = ["I", "m", "9"]

prueba_pw_length = pw_length(10, pw_prueba)

print(prueba_pw_length)
```

⚡`pw_complete()`

1. Recibe los resultados de las funciones anteriores (lista de caracteres basicos, numero de caracteres faltantes y valores True/False)
2. Crea un `pool` con todos los caracteres permitidos según los valores True/False.
3. Selecciona aleatoriamente tantos caracteres del `pool` como se necesite y los añade a la lista original
4. Devuelve una lista con la longitud total establecida en `length`

```python
def pw_complete(pw_basic, pw_faltantes, mayuscula, minuscula, numero, simbolo):

    pool = ""

    if mayuscula:
        pool += string.ascii_uppercase

    if minuscula:
        pool += string.ascii_lowercase

    if numero:
        pool += string.digits

    if simbolo:
        pool += string.punctuation

    azar = np.random.default_rng()

    extras = azar.choice(
        list(pool),
        size=pw_faltantes
    )

    pw_basic.extend(extras.tolist())

    return pw_basic
```

```python
#Prueba de uso
pw_1 = ["I", "m", "9"]
pw_2 = 8


prueba_pw_complete = pw_complete(pw_1, pw_2, True, True, False, True)

print(prueba_pw_complete)
```

⚡`pw_clean()`

1. Recibe una lista con distintos elementos
2. Reordena de forma aleatoria los elementos de la lista
3. Convierte la lista en un `string`
3. Devuelve un `string`

```python
def pw_clean (pw_complete_list):

  azar = np.random.default_rng()

  azar.shuffle(pw_complete_list)

  return "".join(pw_complete_list)
```

```python
#Prueba de uso
pw_complete_01 = ['I', 'm', '9', '<', 'y', 'z', '"', 'I', 'f', '$', '$']

prueba_pw_clean = pw_clean(pw_complete_01)

print(prueba_pw_clean)
```

⚡`main_password()`

1. Recibe de forma externa los resultados obtenidos por `pw_must`
2. Crea una lista de caractares basicos con `pw_elements`
3. Calcula los caracteres faltantes con `pw_length`
4. Completa la lista de caracteres con `pw_complete`
5. Reordena y convierte la lista en `string` con `pw_clean`
6. Devuelve la contraseña final

```python
def main_password(
    length,
    mayuscula=True,
    minuscula=True,
    numero=True,
    simbolo=False):

    if not any((mayuscula, minuscula, numero, simbolo)):
        raise ValueError(
            "Debe seleccionarse al menos un tipo de carácter.")


    pw_basic = pw_elements(
        mayuscula,
        minuscula,
        numero,
        simbolo
    )

    pw_faltantes = pw_length(
        length,
        pw_basic
    )

    pw_list = pw_complete(
        pw_basic,
        pw_faltantes,
        mayuscula,
        minuscula,
        numero,
        simbolo
    )

    password = pw_clean(pw_list)

    print(password)

    return password
```

```python
import pandas as pd
```

⚡`cargar_partidos()`

1. Recibe un archivo .txt
2. Asigna nombre a las columnas
3. Elimina espacios
5. Devuelve un DataFrame


```python
def cargar_partidos (archivo):

  data_partidos = pd.read_csv(archivo, sep="\t", header=None)

  data_partidos.columns = ["Local", "Resultado", "Visitante"]

  limpiar_espacios = ["Local", "Visitante", "Resultado"]

  data_partidos[limpiar_espacios] = (
      data_partidos[limpiar_espacios]
      .apply(lambda x: x.str.strip()))

  return data_partidos
```

⚡`crear_resultado()`

1. Recibe un DataFrame
2. Separa el resultado global en `goles_local` y `goles_visitante`
3. Convierte goles a formato númerico
4. Devuelve un DataFrame actualizado

```python
def crear_resultado(data_partidos):

    columnas_goles = [
        "Goles Local",
        "Goles Visitante"]

    data_partidos[columnas_goles] = (
        data_partidos["Resultado"]
        .str.split("-", expand=True))

    data_partidos[columnas_goles] = (
        data_partidos[columnas_goles]
        .astype(int))

    return data_partidos
```

```python
#Prueba de uso
prueba_data = pd.DataFrame({
    "Local": ["Manlleu", "Vilafranca", "Olot"],
    "Resultado": ["0-1", "4-5", "2-2"],
    "Visitante": ["Granollers", "Terrassa", "Girona FC"]
})

p_crear_resultado = crear_resultado(prueba_data)
print(p_crear_resultado)
```

⚡`crear_estadisticas()`

1. Recibe un DataFrame
2. Crea un diccionario vacío.
3. Recorre cada partido
4. Crea cada equipo si todavía no existe
5. Suma goles a favor y en contra
6. Suma partidos ganados, empatados y perdidos
7. Suma puntos
8. Devuelve un diccionario de diccionarios

```python
def crear_estadisticas(data_partidos):

    estadisticas = {}

    for _, partido in data_partidos.iterrows():

        local = partido["Local"]
        visitante = partido["Visitante"]
        goles_local = partido["Goles Local"]
        goles_visitante = partido["Goles Visitante"]

        if local not in estadisticas:
            estadisticas[local] = {
                "Goles Favor": 0,
                "Goles Contra": 0,
                "Ganados": 0,
                "Empatados": 0,
                "Perdidos": 0,
                "Puntos": 0}

        if visitante not in estadisticas:
            estadisticas[visitante] = {
                "Goles Favor": 0,
                "Goles Contra": 0,
                "Ganados": 0,
                "Empatados": 0,
                "Perdidos": 0,
                "Puntos": 0}

        estadisticas[local]["Goles Favor"] += goles_local
        estadisticas[local]["Goles Contra"] += goles_visitante

        estadisticas[visitante]["Goles Favor"] += goles_visitante
        estadisticas[visitante]["Goles Contra"] += goles_local

        if goles_local > goles_visitante:
            estadisticas[local]["Ganados"] += 1
            estadisticas[visitante]["Perdidos"] += 1
            estadisticas[local]["Puntos"] += 3

        elif goles_local < goles_visitante:
            estadisticas[visitante]["Ganados"] += 1
            estadisticas[local]["Perdidos"] += 1
            estadisticas[visitante]["Puntos"] += 3

        else:
            estadisticas[local]["Empatados"] += 1
            estadisticas[visitante]["Empatados"] += 1
            estadisticas[local]["Puntos"] += 1
            estadisticas[visitante]["Puntos"] += 1

    return estadisticas

```

⚡ `crear_clasificacion()`

1. Recibe el diccionario de estadísticas
2. Lo convierte en un DataFrame
3. Convierte el nombre de cada clave del diccionario en la columna Equipo
4. Ordena los equipos por Puntos de mayor a menor
5. Devuelve el DataFrame de clasificación

```python
def crear_clasificacion (estadisticas):

  data_estadistica = pd.DataFrame.from_dict(estadisticas, orient='index')

  data_estadistica.index.name = "Equipo"

  estadisticas_ordenado = data_estadistica.sort_values(by= "Puntos", ascending=False)

  estadisticas_indice = estadisticas_ordenado.reset_index()

  return estadisticas_indice
```

```python
#Prueba de uso
p_crear_clasificacion = crear_clasificacion(p_crear_estadisticas)

print(p_crear_clasificacion)
```

⚡ `obtener_destacados()`

1. Recibe un DataFrame
2. Hace calculos especificos:
    *   Equipo más goleador
    *   Equipo más goleado
6. Devuelve datos equipos, goles total

```python
def obtener_destacados(estadisticas_indice):

    max_goles = estadisticas_indice["Goles Favor"].max()

    equipos_goleadores = estadisticas_indice[
        estadisticas_indice["Goles Favor"] == max_goles
    ]["Equipo"].tolist()

    max_goles_contra = estadisticas_indice["Goles Contra"].max()

    equipos_goleados = estadisticas_indice[
        estadisticas_indice["Goles Contra"] == max_goles_contra
    ]["Equipo"].tolist()

    return equipos_goleadores, max_goles, equipos_goleados, max_goles_contra
```

⚡ `mostrar_resultados_futbol()`

1. Recibe datos de equipos y total goles
2. Comprueba empates
3. Devuelve la información detallada

```python
def mostrar_resultados_futbol(
    equipos_goleadores,
    max_goles,
    equipos_goleados,
    max_goles_contra,
    estadisticas_indice):

    if len(equipos_goleadores) == 1:
        print(
            f"Máximo goleador: {equipos_goleadores[0]} "
            f"con {max_goles} goles marcados.")
    else:
        print(
            f"Máximos goleadores: {', '.join(equipos_goleadores)} "
            f"con {max_goles} goles marcados cada uno.")

    if len(equipos_goleados) == 1:
        print(
            f"Equipo más goleado: {equipos_goleados[0]} "
            f"con {max_goles_contra} goles recibidos.")
    else:
        print(
            f"Equipos más goleados: {', '.join(equipos_goleados)} "
            f"con {max_goles_contra} goles recibidos cada uno.")

    print("\nClasificación final:")

    for indice, equipo in enumerate(estadisticas_indice["Equipo"], start=1):
        print(f"{indice}. {equipo}")
```

⚡ `main_futbol()`

1. Recibe un archivo .txt con resultados futbolísticos
2. Orquesta todas las funciones creadas
3. Oculta el indice del DataFrame
3. Muestra:
    *   DataFrame con todas las estadísticas y resultados

```python
def main_futbol(archivo):

  lista_partidos = cargar_partidos(archivo)
  resultados_partidos = crear_resultado(lista_partidos)
  estadisticas_partidos = crear_estadisticas(resultados_partidos)
  clasificacion_general = crear_clasificacion(estadisticas_partidos)

  equipos_goleadores, max_goles, equipos_goleados, max_goles_contra = (
        obtener_destacados(clasificacion_general))

  mostrar_resultados_futbol(
      equipos_goleadores,
      max_goles,
      equipos_goleados,
      max_goles_contra,
      clasificacion_general)

  display(clasificacion_general.style.hide(axis="index"))
```
