#### ⚡Definir función `calcular_incremento`

1. Definir función con los datos del archivo politica_salariales que recorrar el dataset empleado por empleado
2. Aplicar la función al dataset principal
3. Revisar valores nulos y posibles errores

```python
## Definir función calcular_incremento
def calcular_incremento(fila):

    if fila["Salari mensual"] < politica_salarial["rules"]["min_salary_threshold"]:
        return 0

    incremento = politica_salarial["base_increment_by_group"][fila["Grup Professional"]]

    if fila["Te_fills"] == True:
        incremento += politica_salarial["children_adjustment"]["has_children"]

    adjustments = [
        ("Rango_edat",       "age_adjustment"),
        ("Nivel_coste_vida", "cost_of_living_adjustment"),
        ("Nivel_transporte", "transport_adjustment")
    ]
    for columna, clave in adjustments:
        if pd.notna(fila[columna]):
            incremento += politica_salarial[clave][fila[columna]]

    return min(incremento, politica_salarial["rules"]["max_total_increment_percent"])
```

```python
## Aplicar la función al dataset principal
encuesta_enriquecida["Incremento_total"] = encuesta_enriquecida.apply(
    calcular_incremento,
    axis=1
)
```

Carrega l'arxiu matriu_distancies.xlsx a pandas de manera que les files i columnes siguin les ciutats i s'eliminin "Las Palmas de Gran Canaria" i "Palma".

Desenvolupa una funció que:

*   Rebi: una ciutat d'origen i una ciutat de destinació.
*   Retorni: l'ordre de les ciutats visitades (ruta) i la distància total recorreguda.

La ruta s'ha de construir seguint una estratègia senzilla: anar sempre a la ciutat no visitada més propera fins a arribar a la destinació.

### 🧠 Análisis y planteamiento del ejercicio

### 🧩 Desarrollo de funciones

#### ⚡ Eliminar datos conflictivos de índice y columnas

```python
## Eliminación de "Las Palmas de Gran Cararia" y "Palmas"
matriz_distancia = matriz_distancia.drop(
    index=["Las Palmas de Gran Canaria", "Palma"],
    columns=["Las Palmas de Gran Canaria", "Palma"]
)
```

```python
## Comprobación de resultados
matriz_distancia.columns
```

#### ⚡ Creación de función `control_kms`

```python
def control_kms(origen, destino):
    origen = origen.strip().title()
    destino = destino.strip().title()

    if origen not in matriz_distancia.index:
        return "La ciudad de origen no existe en la matriz"
    if destino not in matriz_distancia.columns:
        return "La ciudad de destino no existe en la matriz"

    visitadas = []
    ruta = [origen]
    ciudad_actual = origen
    distancia_total = 0

    while ciudad_actual != destino:
        visitadas.append(ciudad_actual)

        distancias = matriz_distancia.loc[ciudad_actual]
        distancias = distancias.drop(visitadas)

        siguiente = distancias.idxmin()

        distancia_total += distancias[siguiente]
        ciudad_actual = siguiente
        ruta.append(ciudad_actual)

    return f"Ciudades en la ruta de viaje: {ruta}", f"Distancia total {distancia_total} kms"
```

```python
## Prueba de uso 1
prueba1_e2 = control_kms("barcelona", "murcia")
display(prueba1_e2)
```

```python
## Prueba de uso 2
prueba2_e2 = control_kms("Xaragoza", "murcia")
display(prueba2_e2)
```

### 🏆 Resultado final

- Se evaluó el cambio de valores NaN a 0 en la diagonal pero se descartó porque `idxmin()` devolvería siempre la misma ciudad al ser 0 el valor mínimo, dificultando el recorrido dentro de la función.
- Se añadió un control mínimo de verificación de parámetros con `strip()` y `title()` para normalizar el input del usuario, asumiendo que la función recibirá texto libre. En caso de que los valores vengan pre-establecidos, esta parte de la función se podría borrar.
