### 🧩 Desarrollo paso a paso

```python
import requests
import pandas as pd
```

```python
# API pública JSONPlaceholder
url_base = "https://jsonplaceholder.typicode.com"

posts = requests.get(f"{url_base}/posts")
users = requests.get(f"{url_base}/users")
todos = requests.get(f"{url_base}/todos")
```

```python
## Cantidad de recursos y status
print("Posts:", len(posts.json()), "- Status:", posts.status_code)
print("Users:", len(users.json()), "- Status:", users.status_code)
print("Todos:", len(todos.json()), "- Status:", todos.status_code)
```

```python
## Peticion con error
prueba_error = requests.get(f"{url_base}/posts/9999")

print("Status:", prueba_error.status_code)
```

```python
## Publicación ficticia
publicacion_ficticia = {
    "title": "Probando APIs",
    "body": "Nada mejor que completar los 3 niveles del Sprint 10!... ",
    "userId": 1
}

respuesta_post = requests.post(
    f"{url_base}/posts",
    json=publicacion_ficticia
)

print(respuesta_post.json())
print("Status:", respuesta_post.status_code)
```

```python
## Petición Delete
prueba_delete = requests.delete(f"{url_base}/posts/1")

print(prueba_delete.json())
print("Status:", prueba_delete.status_code)
```

### 🧩 Desarrollo paso a paso

**API seleccionada: MusicBrainz**

MusicBrainz es una API pública que permite consultar información sobre artistas, álbumes lanzamientos, canciones y otros datos musicales.

<br>

**Endpoints utilizados**

- `/ws/2/artist` que permite buscar artistas.
- `/ws/2/release-group` que permite consultar grupos de lanzamientos, como álbumes o singles.

<br>

**Parámetros útiles**

- `query` indica el criterio de búsqueda.
- `fmt=json` devuelve la respuesta en formato JSON.
- `limit` limita el número de resultados.
- `offset` permite continuar la consulta desde una posición determinada.
- `type` permite filtrar por tipo de lanzamiento, por ejemplo `album`, `single` o `ep`.

```python
## Busqueda de NBID de Madonna
url = "https://musicbrainz.org/ws/2/artist"

params = {
    "query": 'artist:"Madonna"',
    "fmt": "json",
    "limit": 5
}

headers = {
    "User-Agent": "Sprint10Vanessa/1.0"
}

respuesta = requests.get(url, params=params, headers=headers)

print("Status:", respuesta.status_code)

datos = respuesta.json()

for artista in datos["artists"]:
    print(
        artista["name"],
        "-",
        artista.get("country"),
        "-",
        artista["id"]
    )
```

```python
## Conseguir catalogo de albumes
url = "https://musicbrainz.org/ws/2/release-group"

params = {
    "artist": "79239441-bfd5-4981-a70c-55c3f15c1287",
    "type": "album",
    "fmt": "json",
    "limit": 100
}

headers = {
    "User-Agent": "Sprint10Vanessa/1.0"
}

respuesta = requests.get(url, params=params, headers=headers)

print("Status:", respuesta.status_code)

datos = respuesta.json()
```

```python
## Comprobar algunos de los resultados
for album in datos["release-groups"][:5]:
    print(
        album["title"],
        "-",
        album.get("first-release-date")
    )
```

```python
## Creación de DataFrame
df_albumes = pd.DataFrame(datos["release-groups"])

df_albumes.head()
```

# **Nivell 3**: API de Open Data Barcelona

### 🚩 Problema planteado

1. Utilitza l’API d’Open Data BCN per:

    - Cercar un dataset del teu interès mitjançant package_search i package_show. .

2. Dels resultats obtinguts, selecciona’n un que tingui recursos disponibles en CSV o JSON.


3. Amb el resource_id del recurs seleccionat, realitza una consulta mitjançant datastore_search per:

    - Recuperar almenys 100 registres.

4. Converteix els resultats a un DataFrame.


5. Desa el DataFrame en un fitxer .csv.

### 🧩 Desarrollo paso a paso

```python
## Busqueda general de datasets relacionados con "bibliotecas"
url_base = "https://opendata-ajuntament.barcelona.cat/data/api/action"

respuesta = requests.get(
    f"{url_base}/package_search",
    params={
        "q": "biblioteques",
        "rows": 10
    }
)

datos_busqueda = respuesta.json()

print("Status:", respuesta.status_code)

for dataset in datos_busqueda["result"]["results"]:
    print("Título:", dataset["title"])
    print("Name:", dataset["name"])
    print()
```

```python
## Revisión del contenido del dataset seleccionado
dataset_id = "dades-xarxa-biblioteques-catalunya"

respuesta_detalle = requests.get(
    f"{url_base}/package_show",
    params={"id": dataset_id}
)

detalle = respuesta_detalle.json()["result"]

print("Dataset:", detalle["title"])

for recurso in detalle["resources"]:
    print("Nombre:", recurso["name"])
    print("Formato:", recurso["format"])
    print("Resource ID:", recurso["id"])
    print("Datastore activo:", recurso.get("datastore_active"))
    print()
```

```python
## Extraer 100 registros del dataset seleccionado
resource_id = "9dbd5010-970d-4308-89ec-4088a90ea9d8"

respuesta_bibliotecas = requests.get(
    f"{url_base}/datastore_search",
    params={
        "resource_id": resource_id,
        "limit": 100
    }
)

datos_bibliotecas = respuesta_bibliotecas.json()

print("Status:", respuesta_bibliotecas.status_code)
print("Registros recuperados:", len(datos_bibliotecas["result"]["records"]))
```

```python
## Comprobación rápida de datos
datos_bibliotecas["result"]["records"][5]
```

```python
## Creación de DataFrame
df_bibliotecas = pd.DataFrame(
    datos_bibliotecas["result"]["records"]
)

df_bibliotecas.sample(5)
```

```python
## Exportar como CSV
df_bibliotecas.to_csv(
    "bibliotecas_barcelona_2025.csv",
    index=False
)
```
